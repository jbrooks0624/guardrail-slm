# Label audit

Fill `human_label` in `data/interim/label_audit_sample.csv` with `A`–`E`. Agreement is a reported diagnostic, not a build gate.

The CSV is written by `uv run python -m guardrail_slm.data.build --stage labels`.


```python
from collections import Counter
from pathlib import Path

import pandas as pd

path = Path("data/interim/label_audit_sample.csv")
df = pd.read_csv(path, keep_default_na=False)
print(len(df), "rows")
print(df["label"].value_counts())
labeled = df[df["human_label"].isin(list("ABCDE"))]
if len(labeled):
    agree = (labeled["label"] == labeled["human_label"]).mean()
    print(f"agreement on {len(labeled)} rows: {agree:.3f}")
    print(Counter(zip(labeled["label"], labeled["human_label"])))
else:
    print("No human_label values yet.")
```
